```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
top_comment = df.groupby('username').size().reset_index(name='jumblah_komentar').sort_values(by='jumblah_komentar', ascending=False).head(5)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_comment, x='username', y='jumblah_komentar')
plt.title('top 5 username berkomentar terbanyak')
plt.show()
```
