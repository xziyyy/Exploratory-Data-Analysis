# EDA

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
segment_category = df.groupby(['Segment', 'Category'])['Sales'].sum().reset_index()
plt.figure(figsize=(12, 5))
sns.barplot(x='Segment', y='Sales', hue='Category', data=segment_category)
plt.title('Sales by Segment and Category')
plt.show()
```

```python
state_profit = df.groupby('State')['Profit'].sum().reset_index()
plt.figure(figsize=(12, 5))
sns.barplot(x='State', y='Profit', hue='State', data=state_profit)
plt.title('menentukan daerah yg paling banyak profit')
plt.show()
```

```python
customer_sales = df.groupby('Customer Name')['Sales'].sum().head(5).reset_index()
plt.figure(figsize=(12, 5))
sns.barplot(x='Customer Name', y='Sales', hue='Customer Name', data=customer_sales)
plt.title('menentukan customer yg paling banyak beli')
plt.show()
```

```python
corr_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('correlation matrix')
plt.show()
```

## Modeling

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python

columns_to_drop = ['Order ID', 'Order Date', 'Ship Date', 'Customer ID', 'Customer Name', 'Product ID', 'Product Name']

x = df.drop(columns=['Sales'] + columns_to_drop, errors='ignore')

categorical_cols = x.select_dtypes(include=['object']).columns
x = pd.get_dummies(x, columns=categorical_cols, drop_first=True)

y = df['Sales']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## Linear Regression

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

# Select only numerical features from x_train
numerical_features = ['Row ID', 'Postal Code', 'Quantity', 'Discount', 'Profit']
x_train_numeric = x_train[numerical_features]

model.fit(x_train_numeric, y_train)
```

Decision tree

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
from sklearn.tree import DecisionTreeRegressor
model_dt = DecisionTreeRegressor()

# Select only numerical features from x_train
numerical_features = ['Row ID', 'Postal Code', 'Quantity', 'Discount', 'Profit']
x_train_numeric = x_train[numerical_features]

model_dt.fit(x_train_numeric, y_train)
```

```python
pred = {}
numerical_features = ['Row ID', 'Postal Code', 'Quantity', 'Discount', 'Profit']
x_test_numeric = x_test[numerical_features]

pred['lr'] = model.predict(x_test_numeric)
pred['dt'] = model_dt.predict(x_test_numeric)

result = pd.DataFrame({
        'model': ['Linear Regression', 'Decision Tree'],
        'mse': [mean_squared_error(y_test, pred['lr']), mean_squared_error(y_test, pred['dt'])],
        'r2_score': [r2_score(y_test, pred['lr']), r2_score(y_test, pred['dt'])]
})

result
```
